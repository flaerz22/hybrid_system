```python
#!/usr/bin/env python3
"""
Красивое форматирование и подсветка JSON.
Путь к файлу указывается прямо в скрипте (константа JSON_FILE).
"""

import json
import re
import os

# =============================================================================
# КОНФИГУРАЦИЯ — УКАЖИТЕ ФАЙЛ ЗДЕСЬ
# =============================================================================
JSON_FILE = 'data/Сервер/videoevents_02_02_2026_page1.json'   # ← путь к вашему JSON файлу
INDENT = 4                       # размер отступа при форматировании
SAVE_FORMATTED = True            # сохранять ли копию *_formatted.json
# =============================================================================


def colorize(json_str: str) -> str:
    """Подсветка синтаксиса JSON ANSI-цветами в терминале."""
    RESET = "\033[0m"
    KEY = "\033[1;36m"        # голубой жирный — ключи
    STRING = "\033[32m"       # зелёный — строки-значения
    NUMBER = "\033[35m"       # пурпурный — числа
    BOOL_NULL = "\033[1;33m"  # жёлтый — true/false/null
    PUNCT = "\033[90m"        # серый — скобки и запятые

    pattern = re.compile(
        r'("(?:\\.|[^"\\])*")(\s*:)?'          # строка (+ двоеточие => ключ)
        r'|(-?\d+\.?\d*(?:[eE][+-]?\d+)?)'     # числа
        r'|\b(true|false|null)\b'              # булево / null
        r'|([{}\[\],:])'                       # пунктуация
    )

    def repl(m):
        if m.group(1) is not None:
            if m.group(2) is not None:         # строка + ":" => ключ
                return KEY + m.group(1) + RESET + m.group(2)
            return STRING + m.group(1) + RESET
        if m.group(3) is not None:
            return NUMBER + m.group(3) + RESET
        if m.group(4) is not None:
            return BOOL_NULL + m.group(4) + RESET
        if m.group(5) is not None:
            return PUNCT + m.group(5) + RESET
        return m.group(0)

    return pattern.sub(repl, json_str)


def main():
    # Путь берётся из константы; если она пустая — спрашиваем интерактивно
    path = JSON_FILE.strip()
    if not path:
        path = input("📂 Путь к JSON файлу: ").strip().strip('"')

    # === Чтение и валидация ===
    try:
        with open(path, 'r', encoding='utf-8') as f:
            data = json.load(f)
    except FileNotFoundError:
        print(f"❌ Файл не найден: {path}")
        return
    except json.JSONDecodeError as e:
        print(f"❌ Ошибка парсинга JSON: {e}")
        return

    # === Красивое форматирование ===
    # ensure_ascii=False — кириллица остаётся читаемой, не \u043a...
    formatted = json.dumps(data, indent=INDENT, ensure_ascii=False, sort_keys=False)

    print("\n" + "=" * 70)
    print(f"📄 Отформатированный JSON: {path}")
    print("=" * 70)
    print(colorize(formatted))
    print("=" * 70)

    # === Сохранение отформатированной копии ===
    if SAVE_FORMATTED:
        out_path = path.rsplit('.', 1)[0] + '_formatted.json'
        with open(out_path, 'w', encoding='utf-8') as f:
            f.write(formatted + '\n')
        print(f"\n✅ Сохранено: {out_path}")


if __name__ == '__main__':
    main()
```
