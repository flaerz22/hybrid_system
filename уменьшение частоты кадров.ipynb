```python
import os
import shutil
from pathlib import Path

def select_every_nth_photo(input_folder, output_folder, step=3):
    """
    Копирует каждое N-е фото из input_folder в output_folder.
    step=3 означает: берём 1-е, 4-е, 7-е... фото (30 FPS → 10 FPS)
    """
    
    # Создаём выходную папку, если её нет
    output_path = Path(output_folder)
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Поддерживаемые форматы изображений
    image_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.webp'}
    
    # Получаем список всех файлов-изображений
    input_path = Path(input_folder)
    image_files = [
        f for f in input_path.iterdir() 
        if f.is_file() and f.suffix.lower() in image_extensions
    ]
    
    # Сортируем по имени (важно для правильного порядка кадров)
    image_files.sort()
    
    print(f"📂 Входная папка: {input_path}")
    print(f"📂 Выходная папка: {output_path}")
    print(f" Найдено изображений: {len(image_files)}")
    print(f"⚙️  Шаг: {step} (берём каждое {step}-е фото)")
    print(f" Будет скопировано: {len(image_files) // step} фото\n")
    
    # Копируем каждое N-е фото
    copied_count = 0
    for i, image_file in enumerate(image_files):
        if i % step == 0:  # Берём 0, 3, 6, 9...
            dest = output_path / image_file.name
            shutil.copy2(image_file, dest)
            copied_count += 1
            
            # Показываем прогресс каждые 100 файлов
            if copied_count % 100 == 0:
                print(f"   Скопировано: {copied_count} фото")
    
    print(f"\n✅ Готово! Скопировано {copied_count} фото из {len(image_files)}")


if __name__ == '__main__':
    # Настройки
    INPUT_FOLDER = 'data/car_qhd_30fps'      # Папка с исходными кадрами
    OUTPUT_FOLDER = 'data/car_qhd_10fps'  # Папка для результата
    STEP = 3  # Каждое 3-е фото (30 FPS → 10 FPS)
    
    # Запуск
    select_every_nth_photo(INPUT_FOLDER, OUTPUT_FOLDER, STEP)
```
